```python
import matplotlib.pyplot as plt


historical_data['Direction'] = historical_data['Direction'].astype(str).str.upper()

selected = ['BUY', 'SELL', 'CLOSE LONG', 'CLOSE SHORT', 'OPEN LONG', 'OPEN SHORT']
count = historical_data[historical_data['Direction'].isin(selected)]['Direction'].value_counts()

plt.figure(figsize=(10, 10))
plt.pie(count, labels=count.index, autopct='%1.1f%%')
plt.title('Distribution of Direction')
plt.savefig('pie.png')
```
